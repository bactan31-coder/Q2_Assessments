```python
"""Version 2: Added comments and few improvements"""

def add_numbers(num1, num2):
    """Return the sum of num1 and num2."""
    return num1 + num2


def subtract_numbers(num1, num2):
    """Return the difference of num1 and num2."""
    return num1 - num2


def multiply_numbers(num1, num2):
    """Return the product of num1 and num2."""
    return num1 * num2


def divide_numbers(num1, num2):
    """Return the quotient of num1 and num2, or a message if num2 is 0."""
    if num2 == 0:
        return "Cannot divide by zero."
    return num1 / num2


def main():
    # A. Input
    first_number = float(input("Enter first number: "))
    second_number = float(input("Enter second number: "))

    print("\nChoose operation:")
    print("1 - Addition")
    print("2 - Subtraction")
    print("3 - Multiplication")
    print("4 - Division")
    choice = input("Enter choice: ")

    # B. Process: pass arguments to the correct function and store the returned value
    if choice == "1":
        result = add_numbers(first_number, second_number)
    elif choice == "2":
        result = subtract_numbers(first_number, second_number)
    elif choice == "3":
        result = multiply_numbers(first_number, second_number)
    elif choice == "4":
        result = divide_numbers(first_number, second_number)
    else:
        result = "Invalid choice."

    # C. Output
    print("\nResult:", result)


main()
```

1. In my calculator program, I created the functions add_numbers, subtract_numbers, multiply_numbers, divide_numbers, and main.

2. Each of the four calculator functions used the parameters num1 and num2 to receive the two numbers.

3. The arguments passed were first_number and second_number, which are the two numbers the user entered, such as 10 and 5.

4. My program stored the value returned by the chosen function in a variable called result, and then displayed it using the print statement.

5. It is better to divide the program into functions because each function has one clear job, which makes the code easier to read, fix, and reuse than one long block of code.
